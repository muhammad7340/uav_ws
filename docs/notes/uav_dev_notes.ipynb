1. File: drone_custom.gazebo Goal: Stop Gazebo from publishing conflicting TF frames and fix the vertical ground truth offset.

Find this block:

```xml
    <plugin
      filename="gz-sim-odometry-publisher-system"
      name="gz::sim::systems::OdometryPublisher">
      <dimensions>3</dimensions>
      <odom_topic>drone/odom_ground_truth</odom_topic>
      <odom_frame>odom</odom_frame>
      <robot_base_frame>drone/base_footprint</robot_base_frame>
      <tf_topic>/tf</tf_topic>
      <xyz_offset>0 0 -0.5</xyz_offset>
    </plugin>
```
Replace with this:

````python
    <plugin
      filename="gz-sim-odometry-publisher-system"
      name="gz::sim::systems::OdometryPublisher">
      <dimensions>3</dimensions>
      <odom_topic>drone/odom_ground_truth</odom_topic>
      <odom_frame>odom</odom_frame>
      <robot_base_frame>drone/base_footprint</robot_base_frame>
      <xyz_offset>0 0 0</xyz_offset>
    </plugin>
````
